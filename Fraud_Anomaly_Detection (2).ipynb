```python
#importing libraries
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.ensemble import RandomForestClassifier, IsolationForest

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    average_precision_score,
    classification_report
)

import warnings
warnings.filterwarnings("ignore")
```

```python
#uploading the dataset
from google.colab import files

uploaded = files.upload()
```

```python
#reading the dataset
df = pd.read_csv("creditcard.csv")

print("Dataset shape:", df.shape)
df.head()
```

```python
print(df.info())
```

```python
#for supervised learning we have to remove null rows
#Finding the incomplete row
print("Missing values:")
print(df.isnull().sum().sum())
```

```python
#classifying how many fraud and how many are legitimate
#class=0--> legitimate
#class=1--> fraud
print("Class distribution:")
print(df["Class"].value_counts())
```

```python
#Calculating the class percentages
print("Class percentage:")
print(df["Class"].value_counts(normalize=True) * 100)      # this will detect class imbalance
```

```python
#Separating the features and the target
X = df.drop("Class", axis=1)
y = df["Class"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
# now Training/test spliting
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("\ny_train:")
print(y_train.value_counts())

print("\ny_test:")
print(y_test.value_counts())
```

```python
#now Scaling  Time and Amount
scaler = StandardScaler()

X_train = X_train.copy()
X_test = X_test.copy()

X_train[["Time", "Amount"]] = scaler.fit_transform(
    X_train[["Time", "Amount"]]
)

X_test[["Time", "Amount"]] = scaler.transform(
    X_test[["Time", "Amount"]]
)
```

```python
X_train.head()
```

```python
#supervised model
#Training Random Forest
rf_model = RandomForestClassifier(
    n_estimators=100,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train, y_train)

print("Random Forest training completed.")
```

```python
#Generating predictions
y_pred = rf_model.predict(X_test)

y_prob = rf_model.predict_proba(X_test)[:, 1]

print("y_pred:", len(y_pred))
print("y_prob:", len(y_prob))
```

```python
#calculating Precision, Recall, F1 and PR-AUC
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
pr_auc = average_precision_score(y_test, y_prob)

print("Precision :", precision)
print("Recall    :", recall)
print("F1 Score  :", f1)
print("PR-AUC    :", pr_auc)
```

```python
# this is the Classification report
print(classification_report(y_test, y_pred))
```

```python
#unsupervised anomaly detection
#Training Isolation Forest
iso_model = IsolationForest(
    n_estimators=100,
    contamination="auto",
    random_state=42,
    n_jobs=-1
)

iso_model.fit(X_train)

print("Isolation Forest training completed.")
```

```python
#Generating Predictions
iso_raw_pred = iso_model.predict(X_test)

# Isolation Forest:
# -1 = anomaly
#  1 = normal
# Convert to:
#  1 = fraud/anomaly
#  0 = normal

iso_pred = np.where(iso_raw_pred == -1, 1, 0)

print("Isolation Forest predictions generated.")
print(pd.Series(iso_pred).value_counts())
```

```python
#Generating Anomaly Scores
iso_scores = -iso_model.decision_function(X_test)

print("Anomaly scores generated.")
print(iso_scores[:10])
```

```python
#Evaluating Isolation Forest
iso_precision = precision_score(y_test, iso_pred, zero_division=0)
iso_recall = recall_score(y_test, iso_pred, zero_division=0)
iso_f1 = f1_score(y_test, iso_pred, zero_division=0)
iso_pr_auc = average_precision_score(y_test, iso_scores)

print("Isolation Forest")
print("----------------")
print("Precision :", iso_precision)
print("Recall    :", iso_recall)
print("F1 Score  :", iso_f1)
print("PR-AUC    :", iso_pr_auc)
```

```python
#Comparing Both Models
#Creating comparison table
results = pd.DataFrame({
    "Model": [
        "Random Forest",
        "Isolation Forest"
    ],
    "Precision": [
        precision,
        iso_precision
    ],
    "Recall": [
        recall,
        iso_recall
    ],
    "F1 Score": [
        f1,
        iso_f1
    ],
    "PR-AUC": [
        pr_auc,
        iso_pr_auc
    ]
})

results
```

```python
#now saving the results
results.to_csv("model_comparison.csv", index=False)

print("Model comparison saved successfully.")
```
